```python
import ast
def convert(obj):
    L=[]
    for i in ast.literal_eval(obj):
        L.append(i[ 'name' ])
    return L    
        
```

```python
movies['genres']=movies['genres'].apply(convert)
```

```python

def convert3(obj):
    L=[]
    counter=0
    for i in  ast.literal_eval(obj):
        if counter !=3:
           L.append(i['name'])
           counter+=1 
        else:
           break
    return L
```

```python
movies['cast']=movies['cast'].apply(convert3)
```

```python
movies.head()
```

```python
def fetchdirector(obj):
    L=[]
    for i in ast.literal_eval(obj):
        if i['job']=='Director':                   
           L.append(i[ 'name' ])
           break                      
    return L 
```

```python
movies['crew']=movies['crew'].apply(fetchdirector)
```

```python
movies['overview']=movies['overview'].apply(lambda x:x.split())
```

```python
movies.head()
```

```python
movies['genres']=movies['genres'].apply(lambda x:[i.replace(" ","")for i in x])
movies['keywords']=movies['keywords'].apply(lambda x:[i.replace(" ","")for i in x])
movies['cast']=movies['cast'].apply(lambda x:[i.replace(" ","")for i in x])
movies['crew']=movies['crew'].apply(lambda x:[i.replace(" ","")for i in x])
```

```python
movies['tags']=movies['overview']+movies['cast']+movies['crew']+movies['genres']+ movies['keywords']
```

```python
new_df=movies[['movie_id','title','tags']]
```

```python
new_df.loc[:, 'tags'] = new_df['tags'].apply(lambda x: " ".join(x))
```

```python
new_df.head()
```

```python
import nltk
```

```python
nltk.download('stopwords')
```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def stem(text):
    y=[]
    for i in text.split():
        y.append(ps.stem(i))

    return " ".join(y)
```

```python
new_df.loc[:, 'tags'] = new_df['tags'].apply(stem)
```

```python
new_df['tags'][0]
```

```python
new_df.loc[:, 'tags'] = new_df['tags'].apply(lambda x: x.lower())
```

```python
new_df.head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=5000, stop_words='english')
vectors = tfidf.fit_transform(new_df['tags'])
```

```python


# Check if vectors are correctly shaped
print(vectors.shape)

```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity=cosine_similarity(vectors)
```

```python
sorted(list(enumerate(similarity[0])),reverse=True,key=lambda x:x[1])[1:6]
```

```python
def recommend(movie):
    movie_index = movies[movies['title'] == movie].index[0]

    distances = similarity[movie_index]

    ranked = sorted(
        list(enumerate(distances)),
        key=lambda x: x[1],
        reverse=True
    )[1:30]

    # 3 most similar
    top_similar = ranked[:3]

    # 3 still relevant but more diverse
    discovery = ranked[4:7]

    selected = top_similar + discovery

    recommended_movies = []
    recommended_posters = []

    for idx, _ in selected:
        movie_id = movies.iloc[idx].movie_id

        recommended_movies.append(
            movies.iloc[idx].title
        )
        
    recommended_posters.append(movie_id)
        

    return recommended_movies, recommended_posters
```

```python
recommend('Spider-Man')
```
